PROFIT ANALYSIS

MONTHLY SALE ANALYSIS

```python
df["Month"]=df["Order_Date"].dt.month_name()
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()

print(monthly_sales)
```

```python
plt.plot(
monthly_sales.index,
monthly_sales.values,
marker='o',
color="#f20909"
)
plt.xlabel('Month')
plt.ylabel("Sales")
plt.title("Monthly sales trend")
plt.show()
```

Sales ↔ Profit

```python
numeric_cols = [
    "Age",
    "Quantity",
    "Sales",
    "Discount",
    "Profit",
    "Shipping_Days",
    "Customer_Rating"
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(10, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```
